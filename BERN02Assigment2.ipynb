```python
import numpy as np
from matplotlib import pyplot as plt
import pandas as pd
import math
from scipy.special import factorial
from sklearn.linear_model import LinearRegression

X = df['yr'] - 1999
Y = df['count']

best_psum = -np.inf
best_b0 = 0
best_b1 = 0

for beta0 in np.arange(-3.0, 3, 0.05):
  for beta1 in np.arange(-3.0, 3, 0.05):
    lambda_total = np.exp(beta0 + (beta1 * X)) #expected bird count for all years under chosen b0 and b1
    P_log = -lambda_total + Y * np.log(lambda_total) - np.log(factorial(Y))
    df_p = pd.DataFrame(P_log)
    p_sum = P_log.sum() #log total probability
    if p_sum > best_psum:
      best_psum = p_sum
      best_b0 = beta0
      best_b1 = beta1

print(best_psum, best_b0, best_b1)

lambda_total = np.exp(best_b0 + (best_b1 * X))

sample1 = np.random.poisson(lambda_total)
sample2 = np.random.poisson(lambda_total)
sample3 = np.random.poisson(lambda_total)

print(sample1, sample2, sample3)

#save samples
data = {
    'Year': df['yr'],
    'Sample1': sample1,
    'Sample2': sample2,
    'Sample3': sample3}

df_sample = pd.DataFrame(data)

df_sample.to_csv('sample_columns.csv', index=False)
```
